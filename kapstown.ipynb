```python
import pandas as pd

df = pd.read_csv("NIKE_GLOBAL_CATALOGUE.csv")

print("Dataset loaded successfully")
print("Rows and columns:", df.shape)
```

## Step 1. Dataset Overview

At this stage, the dataset is loaded and its dimensions are checked.

```python
df.head()
```

## Step 3. Missing Values Check

Checking for missing values in all columns.

```python
missing_values = df.isnull().sum()

missing_values
```

## Step 4. Duplicate Records Check

Checking for duplicate rows in the dataset.

```python
duplicates = df.duplicated().sum()

print("Duplicate rows:", duplicates)
```

## Step 5. Data Type Transformation

Converting date column into datetime format for future analysis.

```python
df["snapshot_date"] = pd.to_datetime(df["snapshot_date"])

print(df["snapshot_date"].dtype)
```

## Step 6. Removing Unnecessary Columns

Technical columns that are not required for business analysis are removed.

```python
columns_to_drop = [
    "product_url",
    "canonical_url",
    "image_url"
]

df_clean = df.drop(columns=columns_to_drop)

print("New dataset shape:", df_clean.shape)
```

## Step 7. Exporting Clean Dataset

Saving the cleaned dataset for Power BI dashboard development.

```python
df_clean.to_csv("nike_cleaned.csv", index=False)

print("Clean dataset exported successfully.")
```
